```python
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(device)
#Use gpu if available or else cpu 
#for fast processing

#Hyper parameters
block_size = 8
batch_size = 4 #how many block in parellel
```

```python
with open("wizard_of_oz.txt", "r", encoding="utf-8") as f:
    text = f.read()

char = sorted(set(text))

print(char)
```

```python
string_to_int = {ch: i for i, ch in enumerate(char)}
int_to_string = {i: ch for i, ch in enumerate(char)}

def encode(s):
    return [string_to_int[c] for c in s]

    #Or use encode = lambda s:[string_to_int[c] for c in s] 

print(encode('Hello'))

def decode(l):
    return ''.join([int_to_string[i] for i in l])

    # OR use decode = lambda l : [int_to_string[i] for i in l]

print(decode([36, 62, 69, 69, 72]))
encodedval = encode('Kavya')
print(encodedval)
decodedval = decode(encodedval)
print(decodedval)
data = torch.tensor(encode(text),dtype=torch.long)
print(data[:100])
```

```python
#Split dataset into training data and validation data
n = int(0.8*len(data))
train_data = data[:n]
val_data = data[n:]
```

```python

x=train_data[:block_size]
y=train_data[1:block_size+1]

for t in range(block_size):
    context = x[:t+1]
    target = y[t]
    print('when input is ',context,'target is',target)
    
```

```python
#Lets check if numpy'cpu' is faster or torch'cuda' as in gpu is faster
torch_rand1 = torch.rand(100,100,100,100).to(device)
torch_rand2 = torch.rand(100,100,100,100).to(device)

start_time = time.time()
# The '@' operator performs Matrix Multiplication (Dot Product)
res_torch = (torch_rand1 @ torch_rand2)
end_time = time.time()

elapsed_time = end_time - start_time
print(f"PyTorch Matrix Mult Time: {elapsed_time:.8f}")

# 2. NumPy Setup (Element-wise Multiplication)
# Note: To make it a fair comparison, I'm converting the torch tensors to numpy
np_rand1 = torch_rand1.cpu().numpy()
np_rand2 = torch_rand2.cpu().numpy()

start_time = time.time()
# np.multiply performs Element-wise multiplication (much simpler than @)
res_np = np.multiply(np_rand1, np_rand2)
end_time = time.time()

elapsed_time = end_time - start_time
print(f"NumPy Element-wise Mult Time: {elapsed_time:.8f}")
```

```python
#Index 0: Has a 10% chance of being picked.
#Index 1: Has a 90% chance of being picked.
# 50% chance for Index 0 (Heads), 50% chance for Index 1 (Tails)
probabilities = torch.tensor([0.5,0.5])
"""Distribution	Number of Outcomes	Example
Bernoulli	Exactly 2	A single flip of a fair coin.
Binomial	Exactly 2 (repeated)	Flipping a coin 10 times and counting heads.
Multinomial	2 or more (repeated)	Rolling a 6-sided die 10 times, or picking from your [0.1, 0.9] tensor 9 times."""
samples = torch.multinomial(probabilities,num_samples=9,replacement = True)
print(samples)
# Probability of landing a 1 (Tails) is 0.9
p = 0.9
# "Flip" once: Bernoulis
single_flip = torch.bernoulli(torch.tensor([p]))
print(single_flip)
#A Binomial distribution is what you get when you repeat a Bernoulli trial multiple times and count the successes.
# n = number of trials (9 flips)
# p = probability of success (0.9)
binomial_dist = torch.distributions.Binomial(total_count=9, probs=0.5)

success_count = binomial_dist.sample()
print(success_count) 
# Example output: tensor(8.0)
```

```python
#Upper triangular and lower triangular:
out = torch.tril(torch.ones(5,5))
print(out)
out2 = torch.triu(torch.ones(5,5))
print(out2)
#classic "Lego build":common in Transformers and Large Language Models. It is used to create a Look-Ahead Mask
out = torch.zeros(5, 5).masked_fill(torch.tril(torch.ones(5, 5)) == 0, float('-inf'))
out
```

```python
"""The Addition: You add this mask to your "Attention Scores" (the raw numbers that decide which words are important).
The Blackout: Adding -inf to any number makes it -inf. The "legal" words stay the same because you added 0."""
```

```python
#Exp function:
"""The Softmax: When you run Softmax on these numbers, the formula is e^x.
For the zeros: e^0 = 1 (This word gets attention).
For the -infinities: e^{-\infty} = 0 (This word gets zero attention)."""
torch.exp(out)
#Because in attention we want:
"""
allowed positions → keep attention
future positions → probability 0
So the words are in as in 1 and 0 as not to be seen, this is called soft attention masking
""
```

```python
#Example:
import torch
import torch.nn.functional as F
energy = torch.tensor([
    [10.0, 10.0, 10.0,10.0,10.0],
    [10.0, 10.0, 10.0,10.0,10.0],
    [10.0, 10.0, 10.0,10.0,10.0],
    [10.0, 10.0, 10.0,10.0,10.0],
    [10.0, 10.0, 10.0,10.0,10.0]
])
out = out+energy
print(out)

#For probabilities(percentage of relevance):
"""If Word 1 has a probability of 0.70 (70%), the model thinks that word is the most important "clue" for what it's currently doing.
If Word 2 has a probability of 0.01 (1%), the model thinks that word is basically noise and ignores it.

A Real-World ExampleImagine the AI is processing this sentence:
"The robot picked up the battery because it..."When the
AI reaches the word "it", 
the probabilities (Attention Weights) might look like this:  WordProbabilityMeaningrobot0.85The AI is 85% sure "it" 
refers to the robot.
battery 0.10There is a 10% chance"it" refers to the battery.
picked0.03Very low relevance.
the0.02Almost no relevance.
future words0.00Blocked by your -inf mask."""

probs = F.softmax(out,dim=-1)
print(probs)


```
